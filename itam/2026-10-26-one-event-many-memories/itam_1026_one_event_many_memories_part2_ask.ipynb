Part one put the week into seven stores. Now we ask them all
one question, the one Lucy asks on Friday afternoon: which
supplier should she call before Saturday, and why? We build
the stores in one call and write a small reader for each.

```python
# 3.1 build the stores of part one, then one reader per store
from shopstores import *
log, db = build_log(), build_db()
docs, g = build_docs(), build_graph()
fact = late_fact("Fresas del Norte", "2026-10-15")
def ask_log():
    return [e["id"] for e in log if e["at"] >= "2026-10-14"]
def ask_fact():
    f = fact
    return f'{f["about"]} was late {f["late"]} of {f["of"]}'
def ask_rule():
    return RULES[-1][2]
print("log     :", ask_log())
print("fact    :", ask_fact())
print("rule    :", ask_rule().split(" and ")[0])
```

```python
# 3.2 SQL, documents and the graph answer the same question
def ask_sql():
    q = """select s.name, count(*) from deliveries d
           join suppliers s on s.id = d.supplier_id
           where d.late_min > 15
           group by s.name order by 2 desc"""
    return db.execute(q).fetchall()
def ask_docs():
    names = sorted({d["supplier"]["name"] for d in docs})
    def late(n):
        return sum(get(d, "late_min") > 15 for d in docs
                   if get(d, "supplier.name") == n)
    return {n: late(n) for n in names}
def ask_graph():
    u = g.to_undirected()
    return {s: sum(g.nodes[d].get("late", False) for d in u[s])
            for s in u if s.startswith("sup:")}
for name, n in ask_sql(): print("sql     :", name, n)
for name, n in sorted(ask_docs().items()):
    print("docs    :", name, n)
for name, n in ask_graph().items():
    print("graph   :", name[4:], n)
```

SQL and the graph agree that Fresas del Norte was late twice.
The documents split that supplier in two names and report
every supplier at one. The log, the fact and the rule hold
pieces and name no supplier. The vector store answers with
a sentence, not a count, and we see which one next.

```python
# 3.3 the vector store answers with the nearest sentences
from embedrec import embed
import math
vecs = embed(NOTES)
def cos(a, b):
    dot = sum(x * y for x, y in zip(a, b))
    return dot / (math.sqrt(sum(x * x for x in a))
                  * math.sqrt(sum(y * y for y in b)))
SATURDAY = "which supplier should Lucy call before Saturday"
def ask_vec(question=SATURDAY):
    q = embed([question])[0]
    ranked = sorted(((cos(q, v), i) for i, v in enumerate(vecs)),
                    reverse=True)
    return [NOTES[i][:34] for _, i in ranked[:2]]
for line in ask_vec(): print("vector  :", line)
```

Now the failures that no single query shows. A fact can go
stale, two sources can disagree, a schema can drift, and
one of several writes can fail. Each has a small, honest
test. We run all four on the same week.

```python
# 3.4 staleness: the stored fact against what the log says now
now = late_fact("Fresas del Norte", "2026-10-16")
print("stored fact:", fact["late"], "of", fact["of"],
      fact["as_of"])
print("log says   :", now["late"], "of", now["of"], now["as_of"])
print("d6 was on time: the fact needs a new date")
```

```python
# 3.5 contradiction: two sources, one delivery, two stories
slip = dict(id="d5", late_min=130, source="slip")
note = dict(id="d5", late_min=0, source="voice note")
def write(old, new):
    if old["late_min"] != new["late_min"]:
        return "HOLD: ask Lucy", old["late_min"], new["late_min"]
    return "write", new["late_min"]
print(write(slip, note))
```

```python
# 3.6 drift: a column is renamed, each store reacts
rename = "alter table deliveries rename column %s to %s"
db.execute(rename % ("late_min", "delay_min"))
try:
    old = "select count(*) from deliveries where late_min > 15"
    db.execute(old)
except Exception as e:
    print("sql  :", e)
for d in docs: d["delay_min"] = d.pop("late_min")
print("docs :", sum((get(d, "late_min") or 0) > 15 for d in docs),
      "late deliveries, no error")
```

SQL fails loudly when the schema changes. The documents
keep answering, and the answer is quietly wrong. Last, the
writes. A new delivery goes to several stores in a row. If
the third write fails, the first two stay, and the stores
no longer tell the same story.

```python
# 3.7 consistency: a new delivery, and the graph write fails
db2, docs2, g2 = build_db(), build_docs(), build_graph()
def n_sql():
    q = "select count(*) from deliveries"
    return db2.execute(q).fetchone()[0]
def n_graph():
    return sum(g2.nodes[n].get("kind") == "delivery" for n in g2)
def add_delivery(d, fail_at=None):
    steps = [("sql", lambda: db2.execute(
                 "insert into deliveries values(?,?,?,?,?,?)",
                 (d[0], 1, d[2], 0, d[5], 1))),
             ("docs", lambda: docs2.append(dict(id=d[0]))),
             ("graph", lambda: g2.add_edge(d[0], "prod:" + d[2]))]
    for name, step in steps:
        if name == fail_at:
            raise RuntimeError(name + " write failed")
        step()
new = ("d7", "Fresas del Norte", "strawberry", "", "",
       3.5, True, "")
try: add_delivery(new, fail_at="graph")
except RuntimeError as e: print(e)
print("sql:", n_sql(), "| docs:", len(docs2),
      "| graph:", n_graph())
```

```python
# 3.8 a check that compares the stores, and a repair from the log
def counts():
    return (n_sql(), len(docs2), n_graph())
print("counts:", counts(), "-> disagree:", len(set(counts())) > 1)
g2.add_node("d7", kind="delivery", late=False)
g2.add_edge("d7", "prod:strawberry")
print("after the repair:", counts())
```

Seven places to keep a fact is a lot to choose from. In
practice the assistant decides where each new fact goes. We
ask Jev, with a typed question and one criterion per store.
Jev answers with a probability for each. Jev advises, and
our code decides.

```python
# 4.1 a typed question for Jev: which store gets this fact?
from typesafe_sdk import Choice
STORES = {
    "episodic": ("Something happened at a time. "
                 "Keep it in the log."),
    "semantic": "A general fact about a supplier or a flavour.",
    "procedural": "A rule or steps Lucy follows from now on.",
    "sql": "Exact numbers to add up, compare or join.",
    "document": "A whole record with its parts inside it.",
    "graph": "How things connect: who supplied what.",
    "vector": "Free text to find later by its meaning."}
QUESTION = {"store": Choice(
    instructions=("The assistant keeps memories for Lucy's ice "
        "cream shop. A new fact has arrived. Pick the one store "
        "that fits how Lucy will want to ask for it later."),
    criteria=STORES)}
print(len(STORES), "stores to choose from")
```

```python
# 4.2 six new facts, one call to Jev each
from jevrec import ask
FACTS = [
    "Fresas MX and Fresas del Norte are the same supplier.",
    "From now on reject deliveries above 4 degrees.",
    "Marta says her strawberry cone melted outside.",
    "Total cost of strawberries this month: 412 dollars.",
    "The truck was late again.",
    "Lacteos Sur is usually reliable for cream."]
answers = []
for n, fact in enumerate(FACTS, 1):
    state = {"new_fact": fact}
    answers.append(ask(f"s04-jev-{n}", state, QUESTION)["store"])
    top = max(answers[-1], key=answers[-1].get)
    print(n, f"{top:10} {answers[-1][top]:.2f}")
```

Jev gave a probability for every store. Our code does not
just take the biggest one. It writes only when that
probability is at least 70 percent, and below that it asks
Lucy. Here Jev was sure of all six. The first fact, the
supplier with two names, is the closest call.

```python
# 4.3 our code decides: write at 70 percent, else ask Lucy
def decide(p, line=0.70):
    top = max(p, key=p.get)
    return top if p[top] >= line else "ask Lucy"
for n, p in enumerate(answers, 1):
    print(n, decide(p))
print("line at 75 percent, fact 1:", decide(answers[0], 0.75))
```

Now the OpenAI Agents API runs one step of the loop. A
hosted session gets one tool, records, that returns what a
store holds. The model asks the stores it needs and writes
the answer Lucy reads. The tool is a plain JSON spec, and
our own function runs when the session asks for it.

```python
# 4.4 the agent: a model, instructions and one tool
from shopreaders import ask_sql, ask_graph, ask_docs
RECORDS_TOOL = {"type": "function", "name": "records",
    "description": "What one store holds about late deliveries.",
    "parameters": {"type": "object",
        "properties": {"store": {"type": "string",
                       "enum": ["sql", "graph", "document"]}},
        "required": ["store"], "additionalProperties": False}}
request = {"model": "gpt-6-astra",
    "instructions": ("You help Lucy run her ice cream shop. Use "
        "the records tool for the stores you need, then answer "
        "in three short sentences and name your evidence."),
    "tools": [RECORDS_TOOL]}
```

```python
# 4.5 one hosted session answers the Saturday question
from agentsapi import lucy_session
def records(store):
    readers = {"sql": ask_sql, "graph": ask_graph,
               "document": ask_docs}
    return str(readers[store]())
run = lucy_session("s04-saturday", request,
    "Which supplier should Lucy call before Saturday, and why?",
    {"records": records})
import textwrap
print(textwrap.fill(run.final_output, 48))
```

```python
# 4.6 what the session asked, and what our code returned
for call in run.tool_calls:
    print("asked:", call["name"], call["args"])
print("our SQL reader returned:")
for row in ask_sql(): print(" ", row)
```
