One week in Lucy's shop gives us one event to follow. On
Thursday the strawberries came two hours late and warm. The
same afternoon a customer's cone melted. On Friday Lucy
changed her rule. Where we keep that event decides how it
can fail, so we keep it in five places and ask each one.

```python
# 1.1 the week's deliveries, as Lucy's helper typed the slips
from shopdata import DELIVERIES, COMPLAINT, RULES
for d in DELIVERIES:
    print(d[0], f"{d[1]:16}", d[2][:5], d[4][5:], f"{d[5]}C",
          "ok" if d[6] else "NO")
print(COMPLAINT[1], COMPLAINT[2], "-", COMPLAINT[4][:21])
```

First place: an episode log. It keeps what happened and when,
one record per happening. The clock needs care. The slip
says 10:10 on the shop's clock, and a log shared with other
shops must say which clock. We store the time in UTC.

```python
# 1.2 an episode log: one record per happening, in UTC
from datetime import datetime, timedelta, timezone
SHOP = timezone(timedelta(hours=-6))
def utc(text):
    local = datetime.fromisoformat(text).replace(tzinfo=SHOP)
    z = local.astimezone(timezone.utc)
    return z.strftime("%Y-%m-%d %H:%MZ")
log = [dict(id=d[0], kind="delivery", at=utc(d[4]), who=d[1],
            temp=d[5]) for d in DELIVERIES]
log.append(dict(id="c1", kind="complaint", at=utc(COMPLAINT[1]),
                who=COMPLAINT[2]))
print("slip 10:10 on Oct 15 ->", log[4]["at"])
print("slip 16:40 on Oct 15 ->", log[-1]["at"])
```

```python
# 1.3 read it back: by a time window, or by who
window = [e["id"] for e in log
          if "2026-10-15" <= e["at"] < "2026-10-16"]
print("window Oct 15 UTC:", window)
print("by who, Fresas MX:", [e["id"] for e in log
                             if e["who"] == "Fresas MX"])
norte = [e["id"] for e in log if e["who"] == "Fresas del Norte"]
print("by who, Fresas del Norte:", norte)
```

The log never forgets, and that is its trap. The same
delivery can arrive twice, once from the slip and once from
Lucy's voice note. Without a key for the happening, the log
counts it twice. The last read also shows a second trap:
the log knows two names for one supplier.

```python
# 1.4 the same delivery logged twice, then a key stops it
log.append(dict(id="d5", kind="delivery", at=log[4]["at"],
                who="Fresas MX", temp=9.0))
oct15 = [e for e in log if e["kind"] == "delivery"
         and e["at"].startswith("2026-10-15")]
print("records:", len(log), "| deliveries on Oct 15:", len(oct15))
seen, kept = set(), []
for e in log:
    if (e["id"], e["kind"]) not in seen:
        seen.add((e["id"], e["kind"])); kept.append(e)
log = kept
print("after the key (id, kind):", len(log))
```

Second place: a semantic fact. It is derived from the log,
so it can say something no single record says. It is also
a copy, and a copy goes stale. We store the date the fact
was derived, and who derived it.

```python
# 1.5 a derived fact: how often a supplier was late
from shopdata import ALIAS, LATE_MIN, minutes_late
def late_fact(supplier, upto):
    rows = [d for d in DELIVERIES if d[4][:10] <= upto
            and ALIAS.get(d[1], d[1]) == supplier]
    late = sum(minutes_late(d) > LATE_MIN for d in rows)
    return dict(about=supplier, late=late, of=len(rows),
                as_of=upto, by="helper")
fact = late_fact("Fresas del Norte", "2026-10-15")
print(fact["about"], f'late {fact["late"]} of {fact["of"]}')
print("as of", fact["as_of"], "by", fact["by"])
```

```python
# 1.6 Friday's on-time delivery d6 makes the stored fact wrong
now = late_fact("Fresas del Norte", "2026-10-16")
print("stored :", fact["late"], "late of", fact["of"],
      "as of", fact["as_of"])
print("derived:", now["late"], "late of", now["of"],
      "as of", now["as_of"])
same = (fact["late"], fact["of"]) == (now["late"], now["of"])
print("stale:", not same)
```

Third place: procedural memory, the rules Lucy follows. A
rule changes, and the old rule is still the reason for last
week's decision. So a new rule is a new row. The old text
stays, with the dates it was in force.

```python
# 1.7 rules as versions: a change adds a row, never an edit
def rule_on(day):
    return [r for r in RULES if r[1] <= day][-1]
import textwrap
for r in RULES:
    print(r[0], "from", r[1])
    print(textwrap.fill(r[2], 46, initial_indent="  ",
                        subsequent_indent="  "))
print("on Oct 15:", rule_on("2026-10-15")[0], "| on Oct 16:",
      rule_on("2026-10-16")[0])
```

```python
# 1.8 the Thursday delivery against the rule in force that day
d5 = DELIVERIES[4]
limit = {"r1": 5.0, "r2": 4.0}
for r in RULES:
    print(r[0], "would reject d5:", d5[5] > limit[r[0]])
print("rule in force on delivery day:", rule_on(d5[4][:10])[0])
```

Fourth place: SQL, here SQLite from the standard library.
Tables with fixed columns make joins and totals exact. The
price is the schema: the table holds only what we decided
to give a column. A complaint with no delivery_id is stored,
but it joins to nothing.

```python
# 1.9 SQLite: suppliers, deliveries, complaints
import sqlite3
db = sqlite3.connect(":memory:")
db.executescript("""
create table suppliers(id integer primary key, name text unique);
create table deliveries(id text primary key, supplier_id integer,
  product text, late_min integer, temp real, accepted integer);
create table complaints(id text, delivery_id text, text text);""")
NAMES = ("Fresas del Norte", "Lacteos Sur", "Hielos Pacifico")
for name in NAMES:
    db.execute("insert into suppliers(name) values(?)", (name,))
rows = db.execute("select id, name from suppliers")
sid = {n: i for i, n in rows}
for d in DELIVERIES:
    db.execute("insert into deliveries values(?,?,?,?,?,?)",
               (d[0], sid[ALIAS.get(d[1], d[1])], d[2],
                minutes_late(d), d[5], int(d[6])))
db.execute("insert into complaints values('c1', NULL, ?)",
           (COMPLAINT[4],))
print(db.execute("select count(*) from deliveries").fetchone()[0],
      "deliveries stored")
```

```python
# 1.10 the exact answer: late deliveries per supplier
q = """select s.name, count(*) from deliveries d
       join suppliers s on s.id = d.supplier_id
       where d.late_min > ? group by s.name order by 2 desc"""
for row in db.execute(q, (LATE_MIN,)):
    print(row)
print("complaint joins to a delivery:", db.execute(
    "select count(*) from complaints c join deliveries d "
    "on d.id = c.delivery_id").fetchone()[0], "rows")
```

SQL is exact only for what the schema holds. And an agent
that writes the query text itself can be handed a name that
is really a command. We try a name typed as Lucy would
never type it, first built into the text, then as a value.

```python
# 1.11 a name built into the query text, then as a parameter
name = "x' or '1'='1"
unsafe = "select name from suppliers where name = '" + name + "'"
print("built text :", len(db.execute(unsafe).fetchall()), "rows")
safe = "select name from suppliers where name = ?"
found = db.execute(safe, (name,)).fetchall()
print("parameter  :", len(found), "rows")
```

Fifth place: a document store. One document per delivery,
with the complaint nested inside it. We use plain JSON
documents and a small path helper, so it runs anywhere.
The shape is simple to write. The trouble is in the links.

```python
# 1.12 one document per delivery, a complaint nested inside
import json
docs = [dict(id=d[0], supplier=dict(name=d[1]), product=d[2],
             arrived=utc(d[4]), temp=d[5]) for d in DELIVERIES]
docs[4]["complaint"] = dict(by="Marta", text=COMPLAINT[4])
def get(doc, path):
    for key in path.split("."):
        doc = doc.get(key) if isinstance(doc, dict) else None
    return doc
print("keys:", " ".join(docs[4]))
print("d5 complaint:", get(docs[4], "complaint.text")[:30])
```

```python
# 1.13 find by field: the supplier name is a string, not an entity
def find(path, value):
    return [d["id"] for d in docs if get(d, path) == value]
SUP = "supplier.name"
print("Fresas del Norte:", find(SUP, "Fresas del Norte"))
print("Fresas MX       :", find(SUP, "Fresas MX"))
docs[0]["supplier"]["phone"] = "55 0100"
print("phone saved in", [d["id"] for d in docs
                         if get(d, "supplier.phone")], "only")
```

Sixth place: a graph, with NetworkX, the network-science
library. A node is a thing: a supplier, a delivery, a
product, a complaint. An edge is a typed link between two
things. The graph keeps the links the other stores had to
leave out, such as which delivery a complaint is about.

```python
# 2.1 the week as a graph: things are nodes, links are edges
import networkx as nx
from shopdata import DELIVERIES, COMPLAINT, ALIAS
from shopdata import LATE_MIN, minutes_late
g = nx.DiGraph()
for d in DELIVERIES:
    g.add_node(d[0], kind="delivery",
               late=minutes_late(d) > LATE_MIN)
    g.add_edge(d[0], "sup:" + d[1], type="delivered_by")
    g.add_edge(d[0], "prod:" + d[2], type="contains")
g.add_edge("c1", "prod:" + COMPLAINT[3], type="about")
print(g.number_of_nodes(), "nodes,", g.number_of_edges(), "edges")
print("networkx", nx.__version__)
```

```python
# 2.2 walk from the complaint to the supplier, edge by edge
def show(path):
    print(" > ".join(n.split(":")[-1] for n in path))
und = g.to_undirected()
near = und["prod:strawberry"]
print("strawberry:", sorted(d for d in near if d.startswith("d")))
show(nx.shortest_path(und, "c1", "sup:Fresas MX"))
```

The walk ends at a supplier named Fresas MX. The helper also
typed Fresas del Norte on other slips. The graph holds two
nodes for one supplier, and no walk can know they are one.
Entity resolution merges them, once, with a rule we write.

```python
# 2.3 two names, one supplier: merge the alias node
names = sorted(n for n in g if n.startswith("sup:"))
print("supplier nodes:", len(names))
for n in names: print(" ", n[4:])
for alias, real in ALIAS.items():
    g = nx.contracted_nodes(g, "sup:" + real, "sup:" + alias,
                            self_loops=False)
und = g.to_undirected()
print("after the merge:", sum(n.startswith("sup:") for n in g))
one = und["sup:Fresas del Norte"]
print("late deliveries of the one supplier:",
      sorted(d for d in one if g.nodes[d].get("late")))
```

```python
# 2.4 the complaint reaches the supplier through a product
show(nx.shortest_path(und, "c1", "sup:Fresas del Norte"))
print("the complaint is about a product, not a delivery")
```

The walk picked d1, an on-time delivery from last week. It
is a fair path, and the wrong one. We stored the complaint
against a product. We add the missing link with a rule:
the delivery of that product on the same day.

```python
# 2.4b link the complaint to the delivery with a dated rule
for d in DELIVERIES:
    if d[2] == COMPLAINT[3] and d[4][:10] == COMPLAINT[1][:10]:
        g.add_edge("c1", d[0], type="same_day")
und = g.to_undirected()
show(nx.shortest_path(und, "c1", "sup:Fresas del Norte"))
```

A graph invites more edges, and more edges are not more
answers. If we link every pair of deliveries that share a
product, the picture fills in. A person can no longer read
why one supplier was chosen. We count what that costs.

```python
# 2.5 edges grow with the square of the things
print("6 deliveries, every pair linked:",
      nx.complete_graph(6).number_of_edges(), "edges")
print("100 deliveries, every pair linked:",
      nx.complete_graph(100).number_of_edges(), "edges")
```

Seventh place: a vector store. We turn each sentence of
the week into a list of numbers, an embedding, and keep the
lists. A question becomes a list too. The closest lists
are the closest meanings. It finds by meaning and cannot
give an exact answer. The embeddings were made once.

```python
# 2.6 embed the week's sentences once, then replay the numbers
import math
from embedrec import embed
NOTES = [
    "Fresas del Norte strawberries arrived 5 minutes early "
    "at 3.5 C",
    "Fresas del Norte strawberries arrived 45 minutes late "
    "at 4 C",
    "Lacteos Sur cream arrived 80 minutes late at 4 C",
    "Hielos Pacifico cones arrived early but the box was torn",
    "Fresas MX strawberries arrived 130 minutes late at 9 C",
    "Fresas del Norte strawberries arrived 5 minutes late "
    "at 3.5 C",
    "My strawberry cone melted before I got outside.",
    "Rule r1: reject a delivery above 5 degrees C.",
    "Rule r2: reject a delivery above 4 degrees C and write "
    "the arrival time.",
]
vecs = embed(NOTES)
print(len(vecs), "texts,", len(vecs[0]), "numbers each")
```

```python
# 2.7 cosine similarity in plain Python, then ask by meaning
def cos(a, b):
    dot = sum(x * y for x, y in zip(a, b))
    return dot / (math.sqrt(sum(x * x for x in a))
                  * math.sqrt(sum(y * y for y in b)))
def top(question, k=3):
    q = embed([question])[0]
    best = sorted(((cos(q, v), i) for i, v in enumerate(vecs)),
                  reverse=True)[:k]
    for s, i in best:
        print(f"{s:.2f}  {NOTES[i][:36]}")
top("the cone that melted")
```

```python
# 2.8 the same store, a question that needs an exact fact
top("which supplier was late on Thursday?")
print("---")
top("how many times was Fresas del Norte late?")
```
